```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load data from CSV file
data = pd.read_csv('datos.csv')

# Extract t-values (assuming the first column is the time axis)
t = data.iloc[:, 0].values

# Extract all y-values (assuming all columns after the first are Y-values)
y_values = data.iloc[:, 1:].values

# Calculate the average y-values
y_avg = np.mean(y_values, axis=1)

# Plotting
plt.rcParams.update({'font.size': 16})

# Plot each graph
for i, y in enumerate(y_values.T):  # Transpose to iterate over columns
    plt.plot(t, y, label=f'Graph {i + 1}')

# Plot the average graph
plt.plot(t, y_avg, label='Gráfica promedio', linestyle='--', linewidth=2, color='black')

plt.xlabel('Tiempo (s)', fontsize=18)
plt.ylabel('Y-axis', fontsize=18)
plt.legend(fontsize=14)
plt.title('Average of Graphs', fontsize=20)

plt.show()
```
